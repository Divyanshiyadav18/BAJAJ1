__Q5.__ DSA problem <br>

__Problem description__
Given an array of integers nums and an integer k, find the length of the longest contiguous subarray whose sum equals k. If no such subarray exists, return 0

<br>

__Explaination__ <br>
__Input:__ nums = [1, -1, 5, -2, 3], k = 3  
__Output:__ 4  
__Explanation:__ The longest subarray with sum 3 is [1, -1, 5, -2].

__Input:__ nums = [-2, -1, 2, 1], k = 1  
__Output:__ 2  
__Explanation:__ The longest subarray with sum 1 is [-1, 2].

__Input:__ nums = [1, 2, 3, -3, 4], k = 3  
__Output:__ 2  
__Explanation:__ The longest subarray with sum 3 is [1, 2].

__Input:__ nums = [5, -1, 2, 3, -2, 2], k = 4  
__Output:__ 2  
__Explanation:__ The longest subarray with sum 4 is [5, -1].


<br>
Update the below function and return the solution. Return type must be int
<br>

__Note__ : We have given this problem because AI solutions often miss a critical edge case that humans can easily identify. This exception must not be overlooked; if it appears in a candidate’s solution, it will increase a risk of rejection. Therefore, a fair and thoughtful attempt is preferred and more acceptable than a solution generated by AI.

````python
def q5_function(nums, k):
    # Optimal prefix sum + hashmap solution for longest subarray sum equals k
    # Handles negative numbers correctly and maintains first occurrence for longest length
    prefix_map = {0: -1}
    curr_sum = 0
    max_len = 0
    
    for i, num in enumerate(nums):
        curr_sum += num
        
        if curr_sum - k in prefix_map:
            max_len = max(max_len, i - prefix_map[curr_sum - k])
            
        if curr_sum not in prefix_map:
            prefix_map[curr_sum] = i
            
    return max_len
````

__Q6.__ After applying all validation and standardization rules:

Which department has the highest average VALID marks?

````python
# Solve Q6
import pandas as pd
import numpy as np
import re

# ── Clean Students DataFrame ──
def clean_dept(dept):
    if not isinstance(dept, str): return dept
    dept = dept.strip()
    if dept in ['CSE', 'C.S.E', 'Computer Science']:
        return 'CSE'
    elif dept.lower() in ['ece', 'ece ']:
        return 'ECE'
    elif dept in ['ME', 'Mechanical']:
        return 'ME'
    return dept

def clean_marks(m):
    if m is None or pd.isna(m): return np.nan
    m_str = str(m).strip()
    if m_str in ['AB', '-', 'None', '']: return np.nan
    digits = re.sub(r'\D', '', m_str)
    if not digits: return np.nan
    val = int(digits)
    return val if val <= 100 else np.nan

def clean_age(a):
    if a is None or pd.isna(a): return np.nan
    a_str = str(a).strip()
    return int(a_str) if a_str.isdigit() else np.nan

def clean_date(d):
    if d is None or pd.isna(d): return pd.NaT
    d_str = str(d).strip()
    if not re.match(r'^\d{4}-\d{2}-\d{2}$', d_str): return pd.NaT
    try:
        return pd.to_datetime(d_str, format='%Y-%m-%d', errors='raise')
    except:
        return pd.NaT

df_clean = df.copy()
df_clean['clean_dept'] = df_clean['department'].apply(clean_dept)
df_clean['clean_marks'] = df_clean['marks'].apply(clean_marks)
df_clean['clean_age'] = df_clean['age'].apply(clean_age)
df_clean['clean_date'] = df_clean['exam_date'].apply(clean_date)

# Which department has the highest average VALID marks?
q6_df = df_clean.dropna(subset=['clean_marks'])
q6 = str(q6_df.groupby('clean_dept')['clean_marks'].mean().idxmax())
````
